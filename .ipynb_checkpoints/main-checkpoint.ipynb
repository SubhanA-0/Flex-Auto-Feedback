```python
import time
import threading
import tkinter as tk
from tkinter import messagebox
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager

# --- Global Control Flag ---
stop_flag = False
FEEDBACK_URL = "https://flexstudent.nu.edu.pk/Student/CourseFeedback?dump=%2FhkUACDkVqxO5i%2BVLOhIaw%3D%3D" 

# --- Selenium Automation Logic ---
def setup_driver():
    options = webdriver.ChromeOptions()
    driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)
    return driver

def run_automation(username, password, target_index):
    global stop_flag
    driver = None
    try:
        driver = setup_driver()
        print("Opening FLEX Login...")
        driver.get("https://flexstudent.nu.edu.pk/Login")
        
        # --- LOGIN PHASE ---
        if stop_flag: return
        
        # REPLACE IDs with actual FLEX HTML IDs
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.ID, "username_field_id")) # <-- REPLACE ME
        ).send_keys(username)
        
        driver.find_element(By.ID, "password_field_id").send_keys(password) # <-- REPLACE ME
        driver.find_element(By.ID, "login_button_id").click() # <-- REPLACE ME
        
        print("Logged in successfully. Navigating to feedback...")
        time.sleep(5) 
        
        # --- FEEDBACK LOOP ---
        driver.get(FEEDBACK_URL)
        
        while not stop_flag:
            # Find all buttons that say "Give Feedback"
            buttons = driver.find_elements(By.XPATH, "//button[contains(text(), 'Give Feedback')]")
            
            if not buttons:
                print("Done! No more 'Give Feedback' buttons found.")
                break
                
            print(f"Found {len(buttons)} forms remaining. Opening the next one...")
            buttons[0].click()
            time.sleep(3) # Wait for form to load
            
            if stop_flag: break # Check again before filling
            
            # Find the question rows
            question_rows = driver.find_elements(By.CLASS_NAME, "question-row-class") # <-- REPLACE ME
            
            for row in question_rows:
                if stop_flag: break # Abort if user clicked stop mid-form
                
                radios = row.find_elements(By.XPATH, ".//input[@type='radio']")
                if len(radios) > target_index:
                    driver.execute_script("arguments[0].scrollIntoView();", radios[target_index])
                    radios[target_index].click()
                    time.sleep(0.1) # Small pause to look human
                    
            if stop_flag: break # Don't submit if stopped
            
            # Click submit
            driver.find_element(By.ID, "submit_feedback_btn").click() # <-- REPLACE ME
            print("Form submitted. Waiting for page reload...")
            time.sleep(4)

    except Exception as e:
        print(f"An error occurred: {e}")
    finally:
        if stop_flag:
            print("Automation stopped by user.")
        if driver:
            driver.quit() # Safely close browser
        reset_ui() # Re-enable the Launch button

# --- GUI Setup ---
def show_gui():
    global root, submit_btn, stop_btn, roll_entry, pass_entry, rating_var
    
    root = tk.Tk()
    root.title("FLEX Automator")
    root.geometry("400x420")
    root.configure(bg="#ffffff")

    blue_primary = "#0052cc"
    red_stop = "#cc0000"
    font_main = ("Helvetica", 10)

    def launch_thread():
        global stop_flag
        roll = roll_entry.get().strip()
        password = pass_entry.get().strip()
        
        if not roll or not password:
            messagebox.showerror("Error", "Please enter both your Roll Number and Password.")
            return

        rating_map = {
            "Strongly Disagree (Option 1)": 0,
            "Disagree (Option 2)": 1,
            "Neutral (Option 3)": 2,
            "Agree (Option 4)": 3,
            "Strongly Agree (Option 5)": 4
        }
        target_index = rating_map.get(rating_var.get(), 4)
        
        # Update UI state
        stop_flag = False
        submit_btn.config(state="disabled", bg="#cccccc")
        stop_btn.config(state="normal", bg=red_stop)
        
        # Start automation in background thread
        threading.Thread(target=run_automation, args=(roll, password, target_index), daemon=True).start()

    def stop_thread():
        global stop_flag
        print("Stop signal sent. Waiting for script to safely halt...")
        stop_flag = True
        stop_btn.config(state="disabled", text="Stopping...")

    tk.Label(root, text="FLEX Course Feedback Agent", font=("Helvetica", 14, "bold"), bg="#ffffff", fg=blue_primary).pack(pady=20)

    tk.Label(root, text="Roll Number:", font=font_main, bg="#ffffff").pack(anchor="w", padx=50)
    roll_entry = tk.Entry(root, font=font_main, width=32, bg="#f0f4f8", relief="solid", borderwidth=1)
    roll_entry.pack(pady=5)

    tk.Label(root, text="FLEX Password:", font=font_main, bg="#ffffff").pack(anchor="w", padx=50)
    pass_entry = tk.Entry(root, font=font_main, width=32, show="*", bg="#f0f4f8", relief="solid", borderwidth=1)
    pass_entry.pack(pady=5)

    tk.Label(root, text="Fill all forms with:", font=font_main, bg="#ffffff").pack(anchor="w", padx=50)
    rating_var = tk.StringVar(root)
    rating_var.set("Strongly Agree (Option 5)")
    dropdown = tk.OptionMenu(root, rating_var, "Strongly Agree (Option 5)", "Agree (Option 4)", "Neutral (Option 3)", "Disagree (Option 2)", "Strongly Disagree (Option 1)")
    dropdown.config(bg="#ffffff", font=font_main, width=28, relief="solid", borderwidth=1)
    dropdown.pack(pady=10)

    # Buttons Frame
    btn_frame = tk.Frame(root, bg="#ffffff")
    btn_frame.pack(pady=20)

    submit_btn = tk.Button(btn_frame, text="Launch Automation", font=("Helvetica", 10, "bold"), width=18, bg=blue_primary, fg="#ffffff", command=launch_thread, relief="flat")
    submit_btn.grid(row=0, column=0, padx=5)

    stop_btn = tk.Button(btn_frame, text="Stop", font=("Helvetica", 10, "bold"), width=10, bg="#cccccc", fg="#ffffff", command=stop_thread, relief="flat", state="disabled")
    stop_btn.grid(row=0, column=1, padx=5)

    root.mainloop()

def reset_ui():
    # Because Tkinter doesn't like being updated from a background thread, 
    # we use root.after to schedule the UI update safely on the main thread.
    root.after(0, lambda: submit_btn.config(state="normal", bg="#0052cc"))
    root.after(0, lambda: stop_btn.config(state="disabled", bg="#cccccc", text="Stop"))

if __name__ == "__main__":
    show_gui()
```
